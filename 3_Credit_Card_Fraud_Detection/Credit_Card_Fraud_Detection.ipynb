# Problem Statement of the Project

The rapid increase in digital transactions has led to a rise in credit card fraud cases.
Detecting fraudulent transactions is challenging due to the highly imbalanced nature
of the data, where fraudulent cases are much fewer than legitimate ones.

The goal of this project is to develop a machine learning model that can accurately 
identify fraudulent transactions while minimizing false alarms and improving overall transaction security.

# Objective of the Project..

1. to understand and explore the credit card transaction dataset using Exploratory Data Analysis (EDA)
2. to preprocess the data by handling missing values, scaling features, and addressing class imbalance
3. to analyze transaction patterns and identify differences between fraudulent and genuine transactions
4. to build a machine learning classification model to detect fraudulent transactions
5. to evaluate the model using appropriate performance metrics such as precision, recall, F1-score, and confusion matrix
6. to apply techniques like oversampling or undersampling to improve fraud detection performance
7. to gain practical experience in solving real-world imbalanced classification problems using Data Science techniques

# Importing Required Libraries..

```python
import numpy as pd
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report
```

```python
legit = data[data.Class==0]
fraud = data[data.Class==1]
```

```python
legit.head()
```

```python
fraud.head()
```

```python
new_df= legit.sample(n=492)
new_df.head()
```

```python
new_df.shape
```

```python
new_dataset= pd.concat([new_df,fraud], axis=0)
new_dataset.head()
```

```python
new_dataset.shape
```

```python
x= new_dataset.drop(columns=['Class'], axis=1)
y= new_dataset['Class']
```

```python
x_train, x_test, y_train, y_test= train_test_split(x,y, test_size=0.2, random_state=1)
x_train.head()
```

```python
y_train.head()
```

```python
model = LogisticRegression(max_iter=2000)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_train)
print(y_pred)
```

```python
print(type(y_pred))
```

```python
accuracy= accuracy_score(y_pred, y_train)
print(f'model accuracy={accuracy*100:.2f}%')
```

```python
print(classification_report(y_pred, y_train))
```

​In this project, I developed a machine learning model to detect fraudulent transactions within a highly imbalanced financial dataset.
By implementing a Logistic Regression classifier (with max_iter=2000), the model successfully distinguished between 
legitimate transactions and fraudulent activity.
